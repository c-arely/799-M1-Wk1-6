**Research question:** Does regularization improve out-of-sample predictions of marketing outcomes across three independently analyzed datasets? I predict customer-level `ConversionRate` in the primary dataset and campaign-level `Conversions` in the other two. The outcomes are not directly comparable in units.

All preprocessing is fitted on training data only; alpha is selected with five-fold cross-validation on the training set. 

```python
import numpy as np, pandas as pd, os, warnings
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder, PolynomialFeatures
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from statsmodels.stats.outliers_influence import variance_inflation_factor
from IPython.display import display

FILES = {
 'Customer conversion': 'digital_marketing_campaign_dataset.csv',
 'Google Ads': 'GoogleAds_DataAnalytics_Sales_Uncleaned.csv',
 'Product performance': 'marketing_and_product_performance.csv'
}
for label, filename in FILES.items():
    if not os.path.isfile(filename):
        raise FileNotFoundError(f'Missing {filename}.')
print('All three CSV files found.')
```

## 1. EDA and preparation
The datasets differ in structure, missingness, and target definitions. Google Ads currency fields are converted to numeric; missing predictors are imputed within the training pipeline. Variables that leak target information or occur after the conversion event are excluded.

```python
data={name:pd.read_csv(filename) for name,filename in FILES.items()}
for name, df in data.items():
    print('\n',name,'shape:',df.shape,'missing cells:',int(df.isna().sum().sum()),'duplicates:',int(df.duplicated().sum()))
    display(df.head(2))
    display(df.isna().sum().sort_values(ascending=False).head(5).to_frame('missing'))
```

```python
ads=data['Google Ads'].copy()
for col in ['Cost','Sale_Amount']:
    ads[col]=pd.to_numeric(ads[col].astype(str).str.replace(r'[$,]','',regex=True),errors='coerce')
for col in ['Clicks','Impressions','Leads','Conversions','Conversion Rate']:
    ads[col]=pd.to_numeric(ads[col],errors='coerce')
# We deliberately exclude Leads, Conversion Rate and Sale_Amount as outcome-adjacent or downstream variables.
config={
 'Customer conversion':{'frame':data['Customer conversion'],'target':'ConversionRate','numeric':['Age','Income','AdSpend','ClickThroughRate','WebsiteVisits','PagesPerVisit','TimeOnSite','SocialShares','EmailOpens','EmailClicks','PreviousPurchases','LoyaltyPoints'],'categorical':['Gender','CampaignChannel','CampaignType'],'poly':['AdSpend','ClickThroughRate','TimeOnSite']},
 'Google Ads':{'frame':ads,'target':'Conversions','numeric':['Clicks','Impressions','Cost'],'categorical':['Campaign_Name','Location','Device'],'poly':['Clicks','Impressions','Cost']},
 'Product performance':{'frame':data['Product performance'],'target':'Conversions','numeric':['Budget','Clicks','Discount_Level','Bundle_Price'],'categorical':['Subscription_Tier','Common_Keywords'],'poly':['Budget','Clicks','Discount_Level']}
}
for name,c in config.items():
    print(name,'target:',c['target'],'range:',c['frame'][c['target']].min(),c['frame'][c['target']].max())
```

## 2. Regularized regression with training-only cross-validation

Compare unregularized linear regression with Lasso (L1), Ridge (L2), and Elastic Net (L1+L2). The regularization strength `alpha` controls coefficient shrinkage: larger values impose stronger penalties. `l1_ratio` determines the Elastic Net mixture (closer to 1 means more L1). We tune `alpha` over a log-spaced grid and `l1_ratio` over 0.25, 0.5, 0.75. Five-fold CV minimizes RMSE on training folds; test data are used only once for final evaluation. Categorical variables are one-hot encoded, and numeric features are standardized to make penalties comparable.

```python
from sklearn.model_selection import GridSearchCV, KFold
from sklearn.linear_model import Lasso, Ridge, ElasticNet
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.base import clone

def make_encoder():
    try: return OneHotEncoder(handle_unknown='ignore', sparse_output=False)
    except TypeError: return OneHotEncoder(handle_unknown='ignore', sparse=False)

cv=KFold(n_splits=5,shuffle=True,random_state=42)
alpha_grid=np.logspace(-3,2,8)
all_rows=[]; fitted={}
for name,c in config.items():
    cols=c['numeric']+c['categorical']; target=c['target']
    work=c['frame'][cols+[target]].copy()
    work[target]=pd.to_numeric(work[target],errors='coerce')
    work=work.dropna(subset=[target])
    xtr,xte,ytr,yte=train_test_split(work[cols],work[target],test_size=.2,random_state=42)
    prep=ColumnTransformer([
        ('numeric',Pipeline([('impute',SimpleImputer(strategy='median')),('scale',StandardScaler())]),c['numeric']),
        ('categorical',Pipeline([('impute',SimpleImputer(strategy='most_frequent')),('encode',make_encoder())]),c['categorical'])
    ])
    models={
        'Linear baseline':(LinearRegression(),None),
        'Lasso':(Lasso(max_iter=15000,random_state=42),{'model__alpha':alpha_grid}),
        'Ridge':(Ridge(),{'model__alpha':alpha_grid}),
        'Elastic Net':(ElasticNet(max_iter=15000,random_state=42),{'model__alpha':alpha_grid,'model__l1_ratio':[.25,.5,.75]})
    }
    for model_name,(est,grid) in models.items():
        pipe=Pipeline([('prep',clone(prep)),('model',est)])
        if grid:
            search=GridSearchCV(pipe,grid,cv=cv,scoring='neg_root_mean_squared_error',n_jobs=-1,refit=True)
            search.fit(xtr,ytr); fit=search.best_estimator_; params=search.best_params_; cv_rmse=-search.best_score_
        else:
            fit=pipe.fit(xtr,ytr); params={}; cv_rmse=np.nan
        pred=fit.predict(xte)
        coefs=getattr(fit.named_steps['model'],'coef_',np.array([]))
        all_rows.append({'dataset':name,'model':model_name,'train_n':len(ytr),'test_n':len(yte),
            'test_R2':r2_score(yte,pred),'test_RMSE':np.sqrt(mean_squared_error(yte,pred)),
            'test_MAE':mean_absolute_error(yte,pred),'CV_RMSE':cv_rmse,
            'best_alpha':params.get('model__alpha',np.nan),'l1_ratio':params.get('model__l1_ratio',np.nan),
            'nonzero_coefficients':int(np.sum(np.abs(coefs)>1e-8)) if len(coefs) else np.nan})
        fitted[(name,model_name)]=fit
results=pd.DataFrame(all_rows)
display(results.round(4))
```

## 3. Interpretation: model comparisons and coefficient shrinkage
A lower test RMSE indicates smaller typical prediction error in the target's units. R² below zero indicates performance worse than predicting the held-out mean. Lasso can zero out coefficients; Ridge typically shrinks rather than eliminates coefficients; Elastic Net balances the two. The results are descriptive evidence of predictive performance, not proof of a causal marketing effect. Inspect the output rather than assuming a regularized model wins.

```python
summary=results.loc[results.groupby('dataset')['test_RMSE'].idxmin(),['dataset','model','test_R2','test_RMSE','test_MAE','best_alpha','nonzero_coefficients']]
print('Lowest test RMSE within each dataset:')
display(summary.round(4))
print('Lasso / Elastic Net retained coefficients:')
display(results[results['model'].isin(['Lasso','Elastic Net'])][['dataset','model','nonzero_coefficients','best_alpha','l1_ratio']].round(4))
```

## 4. Overfitting, expected/unexpected results, and limitations
Five-fold CV tunes penalty strength without accessing the held-out test set. The training-only preprocessing pipeline prevents leakage of test-set distributional information. Regularization is expected to help if correlated predictors or many weak signals make unpenalized coefficients unstable; if the outcome is weakly related to available features, even the best tuned model can perform poorly. I need to note whether the model's CV RMSE and held-out RMSE are consistent, and whether Lasso or Elastic Net removes many features.

**Limitations:** Google Ads missing values are imputed; this does not establish missingness is random. Conversion counts are nonnegative but linear predictions need not be; count-specific alternatives could be explored in future work. These results are not causal and do not establish how changing ad spend would affect conversions.


## Enhanced Analysis: Regularization, Model Complexity, and Generalization

### Research Question

How does regularization affect predictive accuracy and model complexity across the three digital marketing datasets?

This additional analysis compares the performance of ordinary linear regression, Lasso, Ridge, and Elastic Net using the previously executed model results.

I examine whether regularization reduces prediction error, whether it simplifies the fitted models, and whether the observed differences are large enough to support meaningful conclusions.

Because the three datasets have different target scales, I compare models within each dataset rather than comparing their raw RMSE values across datasets.


```python

# Inspect the existing Week 2 results
print("Available columns:")
print(results.columns.tolist())

# Organize models within each dataset
week2_comparison = results.copy()

# Display all available model results
for dataset_name, group in week2_comparison.groupby("dataset"):
    print(f"\nDataset: {dataset_name}")
    display(group.round(4))

    # Identify the lowest test RMSE
    best = group.loc[group["test_RMSE"].idxmin()]

    print(
        f"Lowest test RMSE: {best['model']} "
        f"({best['test_RMSE']:.4f})"
    )

    # Compare each model against the linear baseline
    baseline = group.loc[
        group["model"] == "Linear baseline",
        "test_RMSE"
    ]

    if not baseline.empty:
        baseline_rmse = baseline.iloc[0]

        comparison = group[["model", "test_RMSE"]].copy()
        comparison["RMSE_change_vs_baseline_%"] = (
            (comparison["test_RMSE"] - baseline_rmse)
            / baseline_rmse * 100
        )

        print("\nPercentage change relative to linear regression:")
        display(comparison.round(4))
```

```python



import matplotlib.pyplot as plt
import numpy as np

for dataset_name, group in week2_comparison.groupby("dataset"):

    baseline_rmse = group.loc[
        group["model"] == "Linear baseline",
        "test_RMSE"
    ].iloc[0]

    comparison = group.copy()

    comparison["RMSE_change_percent"] = (
        (comparison["test_RMSE"] - baseline_rmse)
        / baseline_rmse
    ) * 100

    fig, ax = plt.subplots(figsize=(8, 4))

    bars = ax.bar(
        comparison["model"],
        comparison["RMSE_change_percent"]
    )

    ax.axhline(0, color="black", linewidth=1)

    # Add percentage labels
    for bar, value in zip(
        bars, comparison["RMSE_change_percent"]
    ):
        ax.annotate(
            f"{value:+.4f}%",
            xy=(bar.get_x() + bar.get_width() / 2, value),
            xytext=(0, 7 if value >= 0 else -7),
            textcoords="offset points",
            ha="center",
            va="bottom" if value >= 0 else "top",
            fontsize=9
        )

    # Give percentage labels room to display
    values = comparison["RMSE_change_percent"]
    spread = max(values.max() - values.min(), 0.01)

    ax.set_ylim(
        min(values.min(), 0) - spread * 0.25,
        max(values.max(), 0) + spread * 0.25
    )

    # Format the graph
    ax.set_title(
        f"Week 2: RMSE Change Relative to Baseline\n{dataset_name}"
    )
    ax.set_xlabel("Model")
    ax.set_ylabel("Change in Test RMSE (%)")
    ax.tick_params(axis="x", rotation=25)

    plt.tight_layout()
    plt.show()


```


### Interpretation and Discussion

This analysis compares the predictive performance of linear regression and three regularization methods across the customer conversion, Google Ads, and product performance datasets.

Regularization introduces a penalty for model complexity, which can help prevent unstable coefficient estimates and reduce overfitting. However, its usefulness depends on the relationships present in the data and the selected penalty strength.

In the customer conversion-rate analysis, Lasso reduced all 20 encoded coefficients to zero. This indicates that the selected regularization strength favored a simplified, intercept-only prediction. Although this model achieved a low test RMSE, the result does not establish that the predictors contain no useful information. It shows that the tested Lasso configuration did not retain them.

The additional percentage comparisons help determine whether the best-performing models achieved meaningful improvements over ordinary linear regression or whether their advantages were relatively small.

The broader implication is that regularization can simplify models without necessarily producing substantial gains in predictive performance. Model selection should therefore consider both out-of-sample accuracy and the complexity of the fitted model.



### Cross-Dataset Interpretation: Regularization and Predictive Performance

The comparison across all three datasets demonstrates that regularization substantially affected model complexity but produced only marginal improvements in predictive accuracy.

For the customer conversion dataset, Lasso and Elastic Net reduced all 23 encoded coefficients to zero while improving test RMSE by approximately 0.0261% relative to ordinary linear regression. This suggests that the selected penalties favored intercept-only predictions over the available linear relationships. However, all test R² values remained slightly negative, indicating weak out-of-sample explanatory performance.

For Google Ads, Ridge achieved the lowest test RMSE of 2.2538, representing a 0.0429% improvement over the baseline. Elastic Net retained only four coefficients compared with Ridge's 20, yet achieved a nearly identical test RMSE of 2.2539. This demonstrates that a simpler model can achieve comparable predictive performance, although neither approach explained much of the variation in conversions.

For product performance, Ridge achieved the lowest test RMSE of 291.0527, an improvement of only 0.0008% relative to the baseline. This difference is too small to establish a meaningful practical advantage. All models also produced negative test R² values.

An additional limitation is that Ridge selected alpha = 100 across all three datasets, the largest value in the tested grid. Therefore, the experiment does not establish whether stronger regularization would produce different results.

Overall, the findings suggest that regularization was more effective at controlling model complexity than improving predictive accuracy. The consistency of weak test R² values across three datasets also raises questions about whether the available predictors adequately capture the relationships needed for reliable conversion forecasting.

These results support the broader capstone finding that increasing or constraining model complexity does not necessarily translate into meaningful improvements in generalization.
